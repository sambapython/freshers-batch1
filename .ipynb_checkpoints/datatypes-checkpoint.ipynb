```python
k=[1000,2000,3000]
print("k=",k)#[1000,2000,3000]
m=k
n=k.copy()
k[0]=1
print("k=",k)#[1,2000,3000]
print("m=",m)
print("n=",n)
```

```python
k=[1000,2000,3000]
print("k=",k)#[1000,2000,3000]
m=k
n=k.copy()
k[0]=1
print("k=",k,id(k))#[1,2000,3000]
print("m=",m, id(m))
print("n=",n, id(n))
```

```python
k=[1000,2000,3000]
m=k
n=k.copy()
print("k=",k,id(k))
print("m=",m, id(m))
print("n=",n, id(n))
# for n it create's different indexed object
```

```python
k=[1000,2000,3000]
m=k
n=k.copy()
print("k=",k[0],id(k[0]))
print("m=",m[0], id(m[0]))
print("n=",n[0], id(n[0]))
# for n it create's different indexed object, but the index references pointing to same object
```

```python
from copy import deepcopy
k=[1000,2000,3000]
m=k# reference copy
n=k.copy()# shallow copy
p=deepcopy(k)# deep copy
print("k=",k,id(k))
print("m=",m, id(m))
print("n=",n, id(n))
print("p=",p,id(p))
# for n it create's different indexed object
```

```python
from copy import deepcopy
k=[1000,2000,3000]
m=k# reference copy
n=k.copy()# shallow copy
p=deepcopy(k)# deep copy
print("k=",k[0],id(k[0]))
print("m=",m[0], id(m[0]))
print("n=",n[0], id(n[0]))
print("p=",p[0],id(p[0]))
# for n it create's different indexed object
```

```python
from copy import deepcopy
k=[1000,2000,3000]
m=k# reference copy
n=k.copy()# shallow copy
p=deepcopy(k)# deep copy
k[0]=1
print("k=",k[0],id(k[0]))
print("m=",m[0], id(m[0]))
print("n=",n[0], id(n[0]))
print("p=",p[0],id(p[0]))
# for n it create's different indexed object
```
