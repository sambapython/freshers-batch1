```python
# write a code to findout howmany +VE, -VE
# +ve count=3, -ve count=1
numbers=[10,-20,30,40]
p_c=0
n_c=0
for number in numbers:#number=10,number=-20, number=30,number=40,next?NO
    if number>0:#01>0:True,-20>0:false,30>0:True,40>0:true
        p_c = p_c+1#0+1=1,1+1=2,2+1=3
    else:
        n_c=n_c+1#n_c=0+1=1
print("+VE count=",p_c,"-VE count=",n_c)
```
